```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('my_database.db')
```

```python
#Reading csv to dataframe
df = pd.read_csv('cleaned_dataset.csv')
```

```python
df.to_sql(
    'direct_mail',      # table name
    conn,
    if_exists='replace',  # 'replace', 'append', or 'fail'
    index=False
)
```

```python
pd.read_sql('''
            SELECT
                COUNT(*) AS Total_mailers_sent,
                SUM(response_flag) AS Total_responses,
                SUM(acquisition_flag) AS Total_acquisitions,
                SUM(response_flag)*100.0 /COUNT(*) AS response_rate_pct,
                ROUND(SUM(acquisition_flag)*100.0 /COUNT(*),2) AS acquisition_rate_pct,
                SUM(mailing_cost_usd) AS total_mailing_cost,
                SUM(estimated_monthly_charge_volume_usd) AS total_charge_volume
            FROM
                direct_mail''',conn)
```

```python

pd.read_sql('''
            SELECT
                offer_type,
                ROUND(SUM(response_flag)*100.0 /COUNT(*),2) AS response_rate,
                ROUND(SUM(acquisition_flag)*100.0 /COUNT(*),2) AS acquisition_rate,
                ROUND(AVG(roi),2) AS average_ROI,
                ROUND(SUM(estimated_monthly_charge_volume_usd)) AS total_charge_volume
            FROM
                direct_mail
            GROUP BY
                offer_type''',conn)
```

```python
pd.read_sql('''
            SELECT
                region,
                offer_type,
                SUM(mailing_cost_usd) AS total_mailing_cost
            FROM
                direct_mail
            WHERE
                acquisition_flag=0
            GROUP BY
                region,
                offer_type
            ORDER BY
                total_mailing_cost DESC
            ''',conn)
```

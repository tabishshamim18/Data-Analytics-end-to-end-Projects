## Pareto Analysis

```python
#Sorting the df in descenting order
acquired_customers_df = acquired_customers_df.sort_values(by='estimated_monthly_charge_volume_usd', ascending=False)
```

```python
#calculating cumulative percentage column in the dataframe
acquired_customers_df['cum_pct'] = (acquired_customers_df['estimated_monthly_charge_volume_usd'].cumsum()\
                                    /acquired_customers_df['estimated_monthly_charge_volume_usd'].sum()) * 100
```

```python
customers_80_perc = len(acquired_customers_df[acquired_customers_df['cum_pct'] <= 80])
total_acquired_customers = len(acquired_customers_df)

print(f'Customers Responsible for 80% of Monthly Charge Volume: {customers_80_perc}')
print('')
print(f'Total Acquired Customers: {total_acquired_customers}')
```

```python
print(f"{round(customers_80_perc/total_acquired_customers*100)}% of customers are responsible for 80% of Monthly Charge Volume.")
```

## Time Trend analysis

```python
#Adding year-month column to look at the trend
df['year_month'] = df['mailer_sent_date'].dt.strftime('%Y-%m')
```

```python
summary_df = df.groupby('year_month')\
                .agg(
                    total_mailers_sent = ('mailer_sent_date', 'count'),
                    total_acquisitions = ('acquisition_flag', 'sum'),
                    total_mailing_cost = ('mailing_cost_usd', 'sum')
                ).reset_index()
```

```python
#Calculating monthly acquisition rate
summary_df['acquisition_rate'] = (summary_df['total_acquisitions']/summary_df['total_mailers_sent'])*100
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(specs=[[{"secondary_y": True}]])

fig.add_trace(
    go.Scatter(
        x=summary_df['year_month'],
        y=summary_df['total_mailers_sent'],
        name='Mailers Sent'
    ),
    secondary_y=False
)

fig.add_trace(
    go.Scatter(
        x=summary_df['year_month'],
        y=summary_df['total_acquisitions'],
        name='Acquisitions'
    ),
    secondary_y=False
)

fig.add_trace(
    go.Scatter(
        x=summary_df['year_month'],
        y=summary_df['total_mailing_cost'],
        name='Mailing Cost ($)'
    ),
    secondary_y=True
)

fig.update_layout(title='Mailer Performance by Month')

fig.show()
```

```python
fig = px.line(
    summary_df,
    x='year_month',
    y='acquisition_rate',
    markers=True,
    title='Monthly Acquisition Rate (%)'
)

fig.update_layout(
    width=1500,
    height=600,
    xaxis_title='Month',
    yaxis_title='Acquisition Rate (%)'
)

fig.show()
```
